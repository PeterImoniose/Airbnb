## **2.5. Data Cleaning and Wrangling**

Thorough data preprocessing was carried out to ensure the dataset was consistent, analytically useful, and free of structural errors. Key operations included data type conversions, string cleaning, numeric transformations, JSON parsing, outlier handling, and feature extraction.

Many columns in the listings dataset were stored as strings or mixed formats, requiring conversion into appropriate data types.  
**ID Conversion**  
The listing ID column was converted to integer to ensure consistency and allow for efficient merging and indexing.

**Date Conversion**  
Columns containing dates were converted into datetime objects to support chronological analysis and host-experience calculations.

**Boolean Conversion**  
Columns indicating yes/no host attributes were converted into boolean types to standardise categorical values.

**Parsing Amenities**  
The amenities column contained JSON-like strings representing lists of amenities for each property. These were parsed into actual Python lists using a custom function.  
This transformation allowed for counting amenities and conducting more detailed feature-based analysis.

**Cleaning and Converting Price Values**  
Price values were stored as strings containing currency symbols and commas. These were cleaned and converted into floats.  
This ensured the price variable was usable in statistical analysis and visualisation.




```python
# Remove exact duplicate rows (851 found on inspection - the same listing
# appearing more than once with identical data, not a legitimate repeat listing).
print(f'Duplicate rows before: {df_airbnb_listings.duplicated().sum()}')
df_airbnb_listings = df_airbnb_listings.drop_duplicates().reset_index(drop=True)
print(f'Shape after dropping duplicates: {df_airbnb_listings.shape}')
```

```python
# Convert the 'id' column to integer type.
df_airbnb_listings['id'] = df_airbnb_listings['id'].astype(int)

# Convert the 'last_scraped' column to datetime objects.
df_airbnb_listings['last_scraped'] = pd.to_datetime(df_airbnb_listings['last_scraped'])

# Convert the 'host_since' column to datetime objects.
df_airbnb_listings['host_since'] = pd.to_datetime(df_airbnb_listings['host_since'])

# Convert 'host_is_superhost', 'host_has_profile_pic', and 'host_identity_verified' columns to boolean type.
df_airbnb_listings['host_is_superhost'] = df_airbnb_listings['host_is_superhost'].astype('boolean')
df_airbnb_listings['host_has_profile_pic'] = df_airbnb_listings['host_has_profile_pic'].astype('boolean')
df_airbnb_listings['host_identity_verified'] = df_airbnb_listings['host_identity_verified'].astype('boolean')

# Parse the 'amenities' column from JSON strings into Python lists using the provided parse_amenities function.
def parse_amenities(amenities_string):
    try:
        return json.loads(amenities_string)
    except json.JSONDecodeError:
        return None
    except TypeError:
        return None

df_airbnb_listings['amenities'] = df_airbnb_listings['amenities'].apply(parse_amenities)

# Clean and convert the 'price' column to float by removing '$' and ',' and then casting to float.
df_airbnb_listings['price'] = df_airbnb_listings['price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
df_airbnb_listings['price'] = pd.to_numeric(df_airbnb_listings['price'], errors='coerce')

# Rename the 'price' column to 'Price ($)'.
df_airbnb_listings.rename(columns={'price': 'Price ($)'}, inplace=True)
```

**Handling and dType conversion in Minimum and Maximum Nights**  
Some listings contained unrealistic values in the minimum and maximum nights fields, such as stays exceeding 365 days. These were replaced with the mean of reasonable values (≤ 365).

```python
# Handle outliers in the 'minimum_nights' column by replacing values greater than 365 with the mean of values less than or equal to 365, and then convert the column to integer type.
mean_minimum_nights_filtered = df_airbnb_listings[df_airbnb_listings['minimum_nights'] <= 365]['minimum_nights'].mean()
df_airbnb_listings['minimum_nights'] = df_airbnb_listings['minimum_nights'].apply(lambda x: mean_minimum_nights_filtered if x > 365 else x)
df_airbnb_listings['minimum_nights'] = df_airbnb_listings['minimum_nights'].astype(int)

# Handle outliers in the 'maximum_nights' column by replacing values greater than 365 with the mean of values less than or equal to 365, and then convert the column to integer type.
mean_maximum_nights_filtered = df_airbnb_listings[df_airbnb_listings['maximum_nights'] <= 365]['maximum_nights'].mean()
df_airbnb_listings['maximum_nights'] = df_airbnb_listings['maximum_nights'].apply(lambda x: mean_maximum_nights_filtered if x > 365 else x)
df_airbnb_listings['maximum_nights'] = df_airbnb_listings['maximum_nights'].astype(int)
```

This step prevented extreme values from distorting summary statistics and EDA.

**Extracting Numerical Bathroom Values**  
The bathrooms_text column contained inconsistent descriptive strings such as “1 bath”, “half-bath”, or “private bath”. A custom parser was created to extract numeric values:

```python
# Create a new numerical 'bathrooms_numeric' column by extracting the number of bathrooms from the 'bathrooms_text' column using the provided extract_bathrooms_numeric function.
def extract_bathrooms_numeric(bathrooms_text):
    if pd.isna(bathrooms_text):
        return None

    text = str(bathrooms_text).lower()

    if 'half-bath' in text:
        return 0.5
    elif 'private bath' in text or 'shared bath' in text:
        parts = text.split(' ')
        for part in parts:
            try:
                return float(part)
            except ValueError:
                continue
        return 1.0
    elif 'bath' in text or 'baths' in text:
         parts = text.split(' ')
         for part in parts:
            try:
                return float(part)
            except ValueError:
                continue
         return None

    else:
        return None

df_airbnb_listings['bathrooms_numeric'] = df_airbnb_listings['bathrooms_text'].apply(extract_bathrooms_numeric)
```

The original text column was removed, and the resulting numeric column renamed:

```python
# Drop the original 'bathrooms_text' columns from df_airbnb_listings.
df_airbnb_listings = df_airbnb_listings.drop(columns=['bathrooms_text'])

# Rename the 'bathrooms_numeric' column to 'bathrooms'.
df_airbnb_listings.rename(columns={'bathrooms_numeric': 'bathrooms_compare'}, inplace=True)
```

This produced a clean, numerical bathroom variable suitable for modelling and descriptive analysis.

#### ***2.7.5C. Handling Missing Values for Bedrooms and Bathrooms***

To produce more reliable imputations, missing values for the variables bedrooms, bathrooms, and bathrooms_compare were filled using a group-based imputation strategy. This method leverages more relevant information from listings with similar characteristics.

**Group-Based Imputation**  
Missing values were first imputed using the mean value within each subgroup, defined by the combination of:  
- Property type (e.g., Apartment, House, Guest suite)
- Room type (e.g., Entire home/apt, Private room, Shared room)
    
This approach assumes that listings of the same property and room type tend to have similar structural features. For example, apartments listed as “Entire home/apt” typically have similar bedroom and bathroom configurations.

The following steps were taken:  
*Bedrooms*  
- The mean number of bedrooms was calculated within each property_type × room_type group.
- Missing values were replaced with the corresponding group mean.
- Remaining missing values (if any) were filled using the overall dataset mean.
- The column was converted to integer type for consistency.

*Bathrooms & Bathrooms_compare*
- The same group-based approach was applied to both bathroom-related fields.
- Missing values were filled using Group means and Dataset-wide means (as a fallback)

This method is more accurate than global imputation because it better respects the structural characteristics of the listings and avoids unrealistic replacements.

*Final Check*  
After completing all imputations, a verification step confirmed that no missing values remained in the three processed columns.

### ***3.2.5. Treating outliers***
To address extreme price values while preserving neighbourhood- and property-specific pricing structure, outliers were corrected using a vectorised, group-aware replacement method. Listings were first segmented by neighbourhood, property type, room type, amenity-density bins, and bedroom count. Within each group, Interquartile Range (IQR) thresholds identified unusually high prices, which were then replaced with the mean price of non-outlier observations in the same group.

```python
def replace_outliers_fast(
    df,
    neighbourhood_col='neighbourhood_cleansed',
    property_col='property_type',
    room_col='room_type',
    amenities_col='amenities_count',
    bedrooms_col='bedrooms',
    price_col='Price ($)',
    amenities_bins=[0, 5, 10, 20, 50, 200]
):
    df = df.copy()

    # 1. Create amenities bins
    df['amenities_group'] = pd.cut(
        df[amenities_col], bins=amenities_bins, labels=False, include_lowest=True
    )

    # 2. Define grouping columns
    group_cols = [
        neighbourhood_col,
        property_col,
        room_col,
        'amenities_group',
        bedrooms_col
    ]

    # 3. Compute IQR statistics per group (bulk, vectorised)
    Q1 = df.groupby(group_cols)[price_col].transform(lambda x: x.quantile(0.25))
    Q3 = df.groupby(group_cols)[price_col].transform(lambda x: x.quantile(0.75))
    IQR = Q3 - Q1

    # 4. Compute outlier thresholds
    upper_bound = Q3 + 1.5 * IQR

    # 5. Compute group means excluding outliers
    mask_non_outlier = df[price_col] <= upper_bound
    mean_non_outliers = (
        df[mask_non_outlier]
        .groupby(group_cols)[price_col]
        .transform('mean')
    )

    # Fill missing means (in case group is all outliers)
    mean_non_outliers = mean_non_outliers.fillna(df[price_col].median())

    # 6. Replace outliers in one vectorised operation
    df.loc[df[price_col] > upper_bound, price_col] = mean_non_outliers[df[price_col] > upper_bound]

    return df
```

To further stabilise the pricing distribution, additional outlier filtering was applied within each room type category using the IQR rule. This ensured that unusually high or low prices-often driven by data entry errors or atypical listings-were removed in a structurally consistent way.  

```python
def remove_price_outliers_fast(df, group_cols, price_col):
    df = df.copy()

    # Compute Q1, Q3, IQR for each group
    stats = df.groupby(group_cols)[price_col].agg(['quantile', 'min', 'max'])
    stats = df.groupby(group_cols)[price_col].agg(
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75)
    )
    stats['IQR'] = stats['Q3'] - stats['Q1']
    stats['lower'] = stats['Q1'] - 1.5 * stats['IQR']
    stats['upper'] = stats['Q3'] + 1.5 * stats['IQR']

    # Merge boundaries back onto original df
    df = df.merge(stats[['lower', 'upper']], left_on=group_cols, right_index=True, how='left')

    # Keep only non-outliers
    filtered_df = df[(df[price_col] >= df['lower']) & (df[price_col] <= df['upper'])].copy()

    removed = len(df) - len(filtered_df)
    print(f"Total outliers removed: {removed}")

    # Cleanup
    filtered_df = filtered_df.drop(columns=['lower', 'upper'])

    return filtered_df


# --- Apply the fast function ---
df_airbnb_clean = remove_price_outliers_fast(
    df_airbnb_clean,
    group_cols=['room_type', 'property_type'],
    price_col='Price ($)'
)
```

# **5.0. Conclusion and Recommendations**
## **5.1. Key Findings**
<p align='justify'>The analysis of Airbnb listings in Rio de Janeiro reveals several patterns in how property characteristics, host attributes, and location influence pricing and guest satisfaction:  

* <p align='justify'>Property and room characteristics remain the strongest predictors of price.
Variables such as accommodates, bedrooms, bathrooms, and room type consistently showed the highest correlations with listing price. Neighbourhood-level differences further contributed to price variation, with premium areas such as Leblon, Ipanema, and Barra da Tijuca achieving considerably higher median prices.

* <p align='justify'>Guest satisfaction is generally high and stable across the city, with most ratings concentrated around 4.5-5. Host attributes-such as response rate, acceptance rate, and years of experience- showed weak but positive associations with guest ratings, suggesting professionalism supports satisfaction but does not strongly differentiate listings.

* <p align='justify'>Host status matters.
Superhosts consistently achieved slightly higher review scores compared to non-superhosts, indicating that platform-recognised host quality does translate into marginally better guest outcomes.

* <p align='justify'>Neighbourhood-specific analysis reveals stronger relationships.
For example, in Copacabana, correlations between price and structural features (bedrooms, bathrooms, accommodates) were significantly stronger than in the full dataset. Similarly, when isolating Entire home/apt listings, relationships between variables became clearer and more consistent.

This demonstrates that city-wide correlations mask important local patterns, reinforcing the importance of neighbourhood-level modelling.

## **5.2. Interpretation**
<p align='justify'>These findings indicate that the Airbnb market in Rio de Janeiro is highly segmented by neighbourhood and property type. Tourist-centric areas command a premium not only because of location but also because they tend to offer larger, better-equipped properties. Host behaviour contributes modestly to guest satisfaction, but structural features and location remain the dominant determinants of pricing.

<p align='justify'>The exploratory work also shows that granular segmentation (by neighbourhood or room type) improves model accuracy, suggesting that global models are insufficient for understanding Rio’s diverse and geographically varied Airbnb market.

## **5.3. Limitations**
This study has several limitations that may influence interpretation:

* Temporal limitation: The dataset represents a “snapshot” in time; seasonal demand fluctuations, events, or tourism cycles were not analysed.

* Missing values: Host-related metrics (e.g., acceptance rate) and rating sub-scores include missing entries that could introduce bias.

* <p align='justify'>External factors not included: Crime rates, proximity to beaches or metro stations, and neighbourhood socioeconomic indicators-which are known to influence tourist accommodation markets,were not available in this dataset.

* <p align='justify'>Simplified modelling: Due to time constraints, neighbourhood-specific regression models and advanced predictive tools (e.g., gradient boosting, random forest) were not applied, though early findings suggest they would produce more accurate insights.

## **5.4. Recommendations for Further Analysis**

To enhance the robustness and strategic value of this research, the following directions are recommended:

1. Develop neighbourhood-level predictive models:
As shown in Copacabana and Entire-home analyses, correlations strengthen when the data is segmented. Future work should develop localised hedonic pricing models for major neighbourhoods.

2. Incorporate external geographic variables:
Adding GIS layers such as distance to beaches, metro stations, crime statistics, and tourism hotspots would significantly improve model accuracy.

3. Apply machine learning for price prediction:
Methods such as Random Forest, XGBoost, and Regularised Regression may uncover nonlinear relationships and interactions not evident in simple correlations.

4. Analyse temporal trends:
Including monthly or yearly data would allow measurement of seasonal price patterns, market volatility, and host behaviour over time.
